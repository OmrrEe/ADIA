```python
import pandas as pd
import numpy as np

df = pd.read_csv("sales_data_types.csv")
df.head()
```

```python
df.columns = df.columns.str.strip().str.replace(" ", "_").str.lower()
df.columns
```

```python
df.dtypes
```

```python
df["2016"] + df["2017"]
```

```python
print("""We got the resulting sales for that year, but with the values stored as string, giving us the data as dtype str. We also have the issue where conversion fails and becomes NaN
We fix this by converting the data to the appropiate type, being float, with careful steps to avoid any issues in conversion""")
```

```python
columns_float = ["2016", "2017", "percent_growth", "jan_units"]
columns_boolean = ["active"]

for col in columns_float:
    df[col] = (
        df[col]
        .astype(str)
        .str.replace(r"[^0-9.]", "", regex=True)  # remove $, commas, spaces, etc.
    )
    
    df[col] = pd.to_numeric(df[col], errors="coerce")

for column in columns_boolean: 
  df[column] = df[column].astype(bool)

df.dtypes
```

```python
df["start_date"] = pd.to_datetime(df[["year", "month", "day"]])

df["start_date"]
```
