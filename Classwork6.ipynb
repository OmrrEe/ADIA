## 1. Unit Conversion

Remember that new columns can be generated from simple arithmetic operations on existing columns. For example, if we wanted to make a new column that consists of an old column augmented by 5 units, we can simply code:

```
df["NEW_COL"] = df["COL"] + 5
```

In this case, we want to convert the column "trip_distance" from miles to kilometers into a new column called "trip_distance_km".

We'll use the following unit conversion formula:

$$kilometers = miles \times 1.60934 $$

Try it youself!

````python
### YOUR CODE HERE
df["trip_distance_km"] = df["trip_distance"] * 1.60934

````

## 2. Derived datetime column

Now, we'll extract information about the trip duration, creating a new column called "trip_duration" which contains the amount of <b>minutes</b> the trip lasted.

Remember that when substracting dates, a new <b>timedelta</b> object is created, from which the amount of seconds can be extracted as such:

```
df["NEW_DATETIME_COL"] = (df["DATETIME_COL2"] - df["DATETIME_COL1"]).dt.total_seconds()
```

<b>Hint:</b> Don't forget to divide by 60 to convert seconds into <b>minutes.</b>

````python
### YOUR CODE HERE
df["trip_duration"] = ((df["tpep_dropoff_datetime"] - df["tpep_pickup_datetime"]).dt.total_seconds()) / 60

````

## 4. Creating categorical columns

Let's categorize trips based on their distance. Let's say that a trip that is over 5 km in distance is considered "long" and "short" otherwise.

Create a new column called "length_category" with this information. Remember that you can apply a function to a column in a DataFrame using the .apply method and passing over the name of the pre-defined function. Here's an example:

```
def my_func(value):
  if value == 0:
    return "Zero"
  elif value > 0:
    return "Positive"
  else:
    return "Negative"

df["NEW_COL"] = df["COL"].apply(my_func)
```

I suggest in this case to define a function called "categorize_length" and then apply it to the "trip_distance_km" column.

````python
### YOUR CODE HERE

### Define your function to categorize length here:
def categorize_length(length):
  if length > 5:
    return "long"
  else:
    return "short"
  pass

### Apply your function to create a new column using .apply() here:
df["length_category"] = df["trip_distance_km"].apply(categorize_length)
````

The time of day can be useful to group trips based on whether they happened during the "morning" (from 6 to 11), "afternoon" (12 to 20), "night" (21 to 23) and "late night" (0 to 5).

Create a column called "time_of_day" with this information.

Remmeber to pre-define your function and then apply it!

````python
df.head()
````

````python
### YOUR CODE HERE

### Define your function here:
def categorize_time_of_day(hour):
   if hour >= 0 and hour < 6:
    return "late night"
   elif hour > 5 and hour < 12:
    return "morning"
   elif hour > 11 and hour < 21:
      return "afternoon"
   elif hour > 20 and hour < 24:
      return "night"
pass


### Apply your function to create a new column using .apply() here:
df["time_of_day"] = df["hour_start"].apply(categorize_time_of_day)
````
