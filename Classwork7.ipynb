## Combining multiple conditions

By now, you know how to filter a DataFrame using a single condition. Now we'll combine multiple conditions for filtering.

When combining conditions, it's customary to use the "and" and "or" operators. However, these work for <i>single</i> boolean values. When you're comparing a series of values, you need their bit-wise counterparts & (for "and") and | (for "or").

For example, let's say I wanted to filter products with a price larger than 10, or with a discount < 0.1. I could achieve this with:

```
df[(df["price"]>10) | (df["discount"] <0.1)]
```

Since this operation creates a new DataFrame, I could then chain familiar methods such as .shape, .mean(), etc to get more information.

In this exercise, you need to find out how many trips had an average speed between 2 and 3 (not including either). You can achieve this in two different ways:
1. Filtering the DataFrame and chaining the .shape attribute
2. Combining the masks (boolean conditions) and then chaining the .sum() method.

I highly encourage you to use both methods to verify that you do indeed get the same answer!

Store your answer in a variable called num_speed_2_3.

````python
### YOUR CODE HERE:
num_speed_2_3 = df[(df["average_speed"]>=2) & (df["average_speed"] <= 3)]
### Try with filtering and .shape
num_speed_2_3

````

````python
### Try combining masks and .sum()
mask1 = df["average_speed"]>2
mask2 = df["average_speed"]<3
filtered_df = df[mask1 & mask2]

(mask1 & mask2).sum()
````
