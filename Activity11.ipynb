## What is an Outlier?

Outliers are not the same as errors in the dataset. Outliers are extremely large, extremely small, or generally rare cases in the dataset that are nonetheless real, valid, and not a data collection mistake.
Since they're extreme or rare, they can skew our analysis or model training, which is why classifying and removing them is sometimes employed. There are, however, valid reasons why we'd want to keep outliers!

### Why remove outliers?
- Outliers affect summary statistics like the mean and variance, artificially shifting the data or increasing the variance-leading to skewed analysis in general (hypothesis testing, descriptive statistics, etc)
- Many models can be skewed by outliers, increasing prediction errors and causing problems in generalization since outliers are rare but can be over-represented by these models

### Why keep outliers?
- Outliers are, by definition, valid data! Removing them is removing perfectly valid data that can be studied in its own right!
- Removing outliers can introduce bias in a dataset and artificially reduce the variance
- Outliers can reveal data collection bias in our dataset, for example, if they're from an under-represented population or group
- Outliers can be extremely informative- if a product is selling extremely well, we'd want to explore it and know why

### What to do instead of removing them?
- Get larger datasets so that the outliers don't have a large effect
- Improve data collection to represent outlier cases more fairly
- Use robust models that aren't sensitive to outliers
- Use the median and median absolute deviation instead of the mean and standard deviation, which are less sensitive to outliers
- Analyze outliers separately

Ultimately, it depends on many different factors, and there are trade-offs made all the time. We'll now cover some outlier-detection methods:









## Z-score outlier detection method

 This method works on normally-distributed data and is based on the empirical rule, illustrated below:

![](https://raw.githubusercontent.com/GzzDany/ADIAclasswork/main/empirical_rule.png)

It means that 68% of data are within one standard deviation from the mean, 95% of data are within two standard deviations, and 99.3% of data are within 3 standard deviations.

The normal distribution is common, and arises from the Central-Limit Theorem. Intuitively, whenever there's a phenomenon that's centered around a value (the mean) and has some measure of spread (the standard deviation) and is determined by many factors, the data will be normally-distributed. Here are a couple of examples of normally-distributed data:
- Height of men or women of the same nationality and age group
- Weight of a product from a manufacturing line
- Weight of a bag of chips




The method is simple: Remove all datapoint with a Z-score above 3 or below 3. The Z-score is calculated as follows:
$$Z_i = \frac{x_i-\mu}{\sigma} $$
Where:
* $x_i$ is the individual data point
* $\mu$ is the population mean
* $\sigma$ is the population standard deviation

Here's an intuitive understanding of the Z-score:
If we re-arrange the formula for x, we get
$$ x_i = \mu + Z_i \sigma $$

This literally means that the Z-score represents <b>how many standard deviations the datapoint is away from the mean.</b> In this method, we say that datapoints beyond 3 standard deviations from the mean are outliers.

## MAD outlier detection method

This method is also known as the modified Z-score method. It's based on the same idea as the Z-score, but it uses the median and the median absolute deviation instead of the mean and standard deviation. Here's the formula:

$$M_i = \frac{0.6745(x_i)- median(X)}{MAD} $$
where:
* $MAD = median(|x_i - median(X)|)$, the median absolute deviation
* $median(X)$ is the median of the dataset

Common threshold: Remove values with a M-score greater than 3 (too high) or smaller than 3 (too low).

#### Question: Why is there a 0.6745 in the formula?
Because, if the dataset were normally-distributed, the relationship between the standard deviation $\sigma$ and MAD is: $MAD \approx 0.6745 \sigma$

This way, the MAD and the Z-score behave equally when applied to normally-distributed data.

\


The median is less sensitive to outliers compared to the mean in general. For example, if you were to replace the highest value with some more extreme value, the median would remain the same, while the mean would increase. \
The standard deviation is also an important case here: The standard deviation is the square root of the variance, and the variance is the average of squared differences. If instead of squaring, we take the absolute value, and instead of averaging we get the median, the case is similar - the median absolute deviation is not as sensitive to outliers as the variance or standard deviation.

To diagnose normality, it's a good idea to use histograms and Q-Q plots in conjunction. Histograms can give you an overall feel for the shape of the data, but the number of bins used can bias our analysis. That's where Q-Q plots come in!

### Q-Q plots:
These are plots that show how the distribution of a dataset compares to a theoretical one, in this case, the normal distribution. It plots the data's quantiles against each other, since if two datasets are identically-distributed, their quantiles should match.

We won't get into too much detail for these plots, just know that the closer the graph is to the diagonal line, the better the fit.
- If the graph is above the diagonal, it indicates the data is skewed to the right, meaning the values are much larger than expected if the data followed the normal distribution.
- If the graph is below the diagonal, it indicates the data is skewed to the left, meaning the values are much smaller than expected if the data followed the normal distribution.

## Inter-Quartile Range Outlier Detection

This method is the easiest of all to compute. It relies on the Inter-Quartile Range, and it underlies the logic behind the Box and Whiskers plots we've covered in this class.

The procedure is as follows:
- Calculate the Inter-Quartile Range of the data
- Calculate the upper and lower bounds as

$$Upper bound = median + 1.5\times IQR $$
$$Lower bound = median - 1.5\times IQR $$
- Remove outliers that are beyond the upper and lower boundaries

### Why is there a 1.5?
- Empirical evidence, learned in practice
- Can find outliers without being too strict
- Related to the normal distribution: If the data were normally distributed, then the 1.5 IQR range rule would follow the empirical rule where 99.3% of data are within 3 standard deviations. The IQR is approximately 1.35 times the standard deviation in a normal distribution.
- It's not a rigid choice! It can be adjusted in practice depending on the context of the data - it can be increased when data is very heavily skewed.

Although it's simple, it has a few advantages:
* It makes no assumptions on dataset distribution
* It's even less sensitive to outliers than the MAD method because it relies on the middle 50% of data in total
* It's easy to compute and interpret
