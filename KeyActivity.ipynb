# Question 1: Does delivery speed directly impact customer satisfaction?

* Why is it relevant?
  -  This is a very clear and direct manner of viewing if speed impacts on customer opinions, and it helps to understand if the company should worry more about the speed aspect in their operation or if they can continue operations as normal, mainly, this serves to  understand the relationship between delivery time (difference between purchase and delivery date) and the final review score to then decide if more investment should be implemented into logicsits, like helping identify the max time acceptance on delivery threshhold and could even create a new branch known as "premium" delivery. 

```python
### First I make sure that the necessary data is in datetime columns
df_clean["order_purchase_timestamp"] = pd.to_datetime(df_clean["order_purchase_timestamp"])
df_clean["order_delivered_customer_date"] = pd.to_datetime(df_clean["order_delivered_customer_date"])

# I now filter to have only delivered orders with review scores
df_delivered = df_clean[(df_clean["order_status"] == "delivered") & (df_clean["review_score"].notnull()) & (df_clean["order_delivered_customer_date"].notnull())]

# I then create a custom function to calculate delivery time
def calculate_delivery_days(row):
    return (row["order_delivered_customer_date"] - row["order_purchase_timestamp"]).days

df_delivered["delivery_time_days"] = df_delivered.apply(calculate_delivery_days, axis=1)

# Remove negative or extreme outliers just in case
df_delivered = df_delivered[df_delivered["delivery_time_days"] > 0]
df_delivered = df_delivered[df_delivered["delivery_time_days"] < 62]

# Then just group and aggregate
avg_review_by_delivery = (df_delivered.groupby("delivery_time_days")["review_score"].mean())
```

```python
### YOUR CODE FOR VISUALIZATION HERE

plt.figure(figsize=[10, 6])
plt.plot(avg_review_by_delivery.index, avg_review_by_delivery.values)
plt.xlabel("Delivery Time (Days)")
plt.ylabel("Average Review Score")
plt.title("Average Review Score vs Delivery Time")


plt.show()
```

* Conclusions and observations:

It seems that as delivery time increases, average review scores tend to decrease (with some fluctuations). The general trend suggests that longer delivery times correlate with lower satisfaction, though there are some little spikes of more positive reviews on the days that took longer, probably as these are products with longer delivery time which arrived on time and received their more appropiate review. Still, the general trend suggest that:

- Faster deliveries are associated with higher satisfaction.
- From the drop in positive reviews, it was found that from 20 days and forwards the review scores massively drop in score.

# Question 3: How does revenue change across the seasons?

* Why is it relevant? It helps us understand which seasons are the most profitable, that have the most activity, and which don't. This is useful as we can further adapt the strategy towards the seasons that are most active and have the most activity and profit.

```python
# I first create the value of purchase month, and group each month by order purchase timestamp
df_clean["purchase_month"] = df_clean["order_purchase_timestamp"].dt.month

# Then I apply said group and compare it with the total paid.
monthly_sales = df_clean.groupby("purchase_month")["total_paid"].sum()
```

```python
### YOUR CODE FOR VISUALIZATION HERE

plt.figure()
plt.plot(monthly_sales.index, monthly_sales.values)
plt.xlabel("Month")
plt.ylabel("Total Revenue")
plt.title("Monthly Revenue Trend")
plt.show()
```

* Conclusions and observations. When thinking of sales across seasons and such, one would think that it would be the opposite between the first and last months. When the holidays are coming and everyone is trying to pay things all over the place, I expected there to be a bigger spike on the 11 and 12 months, and yet on the 12th month we have a huge spike downwards, while when it comes back around tot he first and second months it goes back up way more, which for example I thought that now that the holidays are over and such and people would have spent a good amount, the sales would go down. This does help to know which seasons are the most active and have the most opportunity to generate revenue for the bank and which seasons are the weakest and least filled with activity.
