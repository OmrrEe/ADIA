## Frequencies of data entries

The .value_counts() method in pandas is a powerful tool for analyzing the distribution of values in a DataFrame column. It returns a Series containing unique values from the column as the index and their corresponding counts as values, sorted in descending order by default.

It's a powerful way to understand the structure of large datasets.

Since it's applied to a specific column, you must select it and then call the method, as such:

```
df[COLUMN_NAME].value_counts()
```
