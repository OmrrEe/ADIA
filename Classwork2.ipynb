# Creating DateTime objects

We'll need the module "datetime" to work with. This one is part of the Python standard library, so it's automatically installed with the Python language.

Let's load it using the import keyword and its standard alias, dt:

```python
import datetime as dt
```

Now let's create out datetime object. The Titanic struck an iceberg that caused its sinking a few hours later. That occurred on April 14, 1912 at 23:40. Create a datetime object called titanic_date with this specific value.

```python
titanic_sink = dt.datetime(1912, 4, 14, 23, 40)
```

Dates have multiple formats used all around the world. The most common one is ISO 8601 Format which shows the datetime as:

"YYYY-MM-DDTHH:mm:ss"

So for the titanic datetime, it would show it as: \
1912-04-14T23:40:00

To show the datetime in ISO format, use the method dt.isoformat(), its name is short for "String Format Time". Run the following cell:

```python
titanic_sink.isoformat()
```

We can create our own custom datetime format using the following format codes
* Year (in 4-digit format) is represented as %Y
* Year (in 2-digit format) is represented as %y
* Month number is represented as %m
* Month name is represented as %B
* Day of the month is represented as %d

You can find many more codes in the documentation for [Python datetime format codes](https://docs.python.org/3/library/datetime.html#strftime-and-strptime-format-codes).

When we want to show the date as our own custom format, we need to call the .strftime() method and provide our desired format as such:

```python
titanic_sink.strftime("%Y-%m-%d at %H:%M")
```

Notice that we can include any text we want, and only the specific parts using the datetime codes are replaced with their appropriate values. You can get very creative with this!

# Parsing Datetimes

Usually, we have a dataset with datetimes already in text form in some format and we want to manipulate those, instead of already knowing the year, month, day, hour, minute of a given timestamp. So it's the reverse of what we've been doing so far - going from a date in text to a datetime object we can manipulate in Python.

For this, we need to <b>parse</b> dates, which simply means we need to the able to read them in their given format. We use the same codes as before to detect how we can extract the date from text.

For example: The Hindenburg explosion ocurred on May 6, 1937. Suppose we have the text "06/05/1937 18:45". We can convert it into a datetime object using the the format code "%d/%m/%Y %H:%M", but now, instead of using str<b>f</b>time, we use str<b>p</b>time, short for "String Parse Time".

We need to call it from the dt.datetime module and it takes two arguments: The text we want to parse, and the date format.

Run the cell below to create the hindenburg disaster datetime object.

```python
hindenburg_disaster = dt.datetime.strptime("06/05/1937 18:45", "%d/%m/%Y %H:%M")
hindenburg_disaster
```

# Date differences

Now let's make some arithmetic with dates. The question is, how much time passed between the Titanic and the Hindenburg disasters?

In mathematics, the concept of <i>subtraction</i> and <i>difference</i> are often used interchangeably, and with good reason. Here, we can also examine the time difference between two events using the subtraction symbol.

Run the cell below to create the difference object.

```python
difference = hindenburg_disaster - titanic_sink

difference
```

Note that the object created is called a <b>timedelta</b>. It automatically calculates the difference in days with seconds as a remainder, meaning there were 9,154 days and 68,700 seconds between both disasters.

We can extract how many seconds passed between both disasters using the .total_seconds() method.

Run the cell below to see how it would work:

```python
difference.total_seconds()
```
