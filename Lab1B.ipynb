```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("iris.csv")

df.head()
```

```python
df.dtypes
```

```python
df["species"] = df["species"].astype("string")
df.dtypes
```

```python
sns.boxplot(data=df)
plt.title("Boxplot of All Numerical Variables")
plt.show()
```

```python
sns.boxplot(x="species", y="sepal_width", data=df)
plt.title("Sepal Width by Species")
plt.show()
```

```python
grouped = df.groupby("species")["sepal_width"]

Q1 = grouped.quantile(0.25)
Q3 = grouped.quantile(0.75)
median = grouped.median()
IQR = Q3 - Q1

lower_whisker = Q1 - 1.5 * IQR
upper_whisker = Q3 + 1.5 * IQR

stats = pd.DataFrame({"Q1": Q1, "Median": median, "Q3": Q3, "Lower Whisker": lower_whisker, "Upper Whisker": upper_whisker})

stats
```

```python
sns.boxplot(x="species", y="sepal_width", data=df)
sns.swarmplot(x="species", y="sepal_width", data=df, color="black")

plt.title("Sepal Width with Swarmplot")
plt.show()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(10, 8))

sns.boxplot(ax=axes[0,0], x="species", y="sepal_length", data=df)
sns.boxplot(ax=axes[0,1], x="species", y="sepal_width", data=df)
sns.boxplot(ax=axes[1,0], x="species", y="petal_length", data=df)
sns.boxplot(ax=axes[1,1], x="species", y="petal_width", data=df)

plt.tight_layout()
plt.show()
```

```python
print("""Petal length and petal width clearly separate the species.
We can see that Setosa is completely distinct. Versicolor and Virginica are also well separated""")
```

```python
print("""Sepal width and Petal width show visible outliers.
Sepal mainly in Iris-virginica with strange low values, Petal width has some outliers in Iris-setosa.
Petal length/Sepal length don't really have outliers.""")
```
