```python
import numpy as np
import matplotlib.pyplot as plt

X_or = np.array([ #The perceptron learns from this data
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

y_or = np.array([0, 1, 1, 1])
```

```python
class Perceptron:
    def __init__(self, learning_rate=0.1, epochs=20):
        self.lr = learning_rate
        self.epochs = epochs
        self.weights = None
        self.bias = None
        self.errors_per_epoch = []

    def predict(self, X):
        linear_output = np.dot(X, self.weights) + self.bias
        return np.where(linear_output >= 0, 1, 0) #implements prediction using step activation

    def fit(self, X, y):
        n_samples, n_features = X.shape
        self.weights = np.random.rand(n_features) #random weights and bias
        self.bias = np.random.rand()

        for _ in range(self.epochs):
            errors = 0
            for xi, target in zip(X, y):
                linear_output = np.dot(xi, self.weights) + self.bias
                y_pred = 1 if linear_output >= 0 else 0
                update = self.lr * (target - y_pred) #training method with perceptron learning rule
                self.weights += update * xi
                self.bias += update
                errors += int(update != 0)
            self.errors_per_epoch.append(errors)
```

Testing the perceptron

```python
import numpy as np

#In this part, we prepare the training data

# The Inputs (X)
X_and = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

# The Outputs (y)
y_and = np.array([0, 0, 0, 1])

#In this part, we create and test the perceptron

perceptron = Perceptron(learning_rate=0.1, epochs=20)
perceptron.fit(X_and, y_and)

#In this part, we test the predictions

predictions = perceptron.predict(X_and)

print("Predictions:", predictions)
print("Actual:", y_and)
```

Explaining the Perceptron

```python
print("1. INITIALIZATION:")
print("""The perceptron starts with random weights and a bias. For example:
w1 = 0.2, w2 = -0.3, bias = 0.1""")
```

```python
print("2. PREDICTION PROCESS:")
print("""The perceptron takes inputs (x1, x2) and performs a weighted sum:
weighted_sum = (w1 * x1) + (w2 * x2) + bias
Then it applies an activation function (step function):
If weighted_sum >= 0 then output = 1
If weighted_sum < 0 then output = 0
      
So for an example, with inputs (1, 1), we get:
weighted_sum = (0.2 * 1) + (-0.3 * 1) + 0.1 = 0.0
our output here = 1 (since it's >= 0)""")
```

```python
print("3. LEARNING PROCESS:")
print("""The perceptron compares its prediction with the actual output (target).
For the AND function, the correct outputs are (0,0 becomes 0), (0,1 becomes 0), (1,0 becomes 0), (1,1 becomes 1).

The error is calculated as: error = target - prediction. 
If there is an error, the weights and bias are updated using:
w = w + (learning_rate * error * input) and bias = bias + (learning_rate * error). 

This process repeats over many iterations (epochs), allowing the perceptron to gradually adjust and improve its predictions.""")
```

```python
print("4. OUTCOME:")
print("""After multiple iterations (epochs), the perceptron learns the correct pattern:
Final behavior:
Only outputs 1 when BOTH inputs are 1. Otherwise, it outputs 0.
This means the perceptron has successfully learned the AND logic function and that 
the model has successfully adjusted its weights and bias to correctly classify the data.""")
```
