# Creating a pandas Series

Remember: A pandas Series is a one-dimensional array of data with a labeled index.

The data can be of various types, check out the table below:


| Usage | Pandas dtype | Python Type | Example |
| :--| :-- | :-- | :-- |
| Text | object | str or mixed | names, countries, IDs, etc|
| Integers | int64 | int | Whole numbers: -10, 2, 60, 420 |
| Numeric with decimals | float64 | float | Numbers with decimals: 3.1416, 10.1 |
| True or False | bool | bool | Two possible values: True or False|
| Date | datetime64 | NA | Date and time values: 25-02-2024 13:00:00


You can create a Series from a list of values, as an example, let's create a series of fruit names. Run the cell below to see this in action!

````python
fruits = pd.Series(["apple", "orange", "banana"], name="Fruits")

fruits
````

````python
series = pd.Series([1, 2, 3.7, "one"], name="Test")

series
````

In the code above, we create a Pandas Series with 3 values (apple, orange and banana), and store it in the variable called "fruits".

You can see the index starts from 0 and increases by 1 for each element in the Series, the name displayed at the top and the data type "object" displayed at the bottom, indicating that the series contains text.

When you create a pandas.Series, pandas automatically determines the data type (dtype) based on the values you provide. Since the values in this case were text, pandas sets the data type to "object", which is used for string-based data.

It's important to note that <b>all values in a Series must have the same data type. </b>If you mix different types, pandas will convert them to a common type that can accommodate all values without loss of information.

For example:
* Mixing integers and floats results in all values converted to <b>float</b>, since integers can be represented as floats by adding ".0", but floats cannot be represented as integers without losing information.
* However, if you mix numbers and text, pandas will convert everything to "object" (string format), since text cannot be represented as numeric type.

Data types are an important aspect to consider when storing information both for efficiency and functionality. The data type determines what kind of transformations can be performed on the data later on - such as adding a constant, or multiplying all values by a constant, which you cannot do with text.

# Creating DataFrames

A pandas DataFrame is, informally, a table of data with an index and column names.

One way to create a DataFrame is to use a dictionary. You'll cover these useful and versatile data structures in detail in the Python programming module later. For now, just know that a dictionary is defined with curly braces { } and has keys and values associated with a colon (:). In this case, the keys will be column names, and the values will be lists of the values in the column, as shown below:

````python
data_dictionary = {"Fruit":["apple", "orange", "banana"], "Price":[1.2, 0.95, 2.3]}


fruit_prices = pd.DataFrame(data=data_dictionary)

fruit_prices
````

Notice the index at the far left, and the column names on the top of the DataFrame object.

To recap, the code above created a dictionary with two keys: "Fruits" and "Prices", and their associated values are the lists of fruit names and prices, respectively. You can also avoid the step of creating the dictionary outside and type it directly into the pd.DataFrame function as such:

```
pd.DataFrame(data={"Fruits":["apple", "orange", "banana"], "Prices":[1.20, 0.95, 2.30]})
```


If you want to inspect the data types, you can call the DataFrame as shown below. Run the cell below to show the output of the code.

````python
fruit_prices.dtypes
````

Notice that each column name is now in a row and its associated value is the datatype of that column.
