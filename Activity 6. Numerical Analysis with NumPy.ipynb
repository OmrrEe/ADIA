```python
import numpy as np

data = np.genfromtxt("DatosAutos.csv", delimiter=",", skip_header=1)

year = data[:,0]
price = data[:,1]
trans = data[:,2]
mileage = data[:,3]
fuel = data[:,4]
tax = data[:,5]
mpg = data[:,6]
engineSize = data[:,7]

# first phase
print("Max tax:", np.max(tax))
print("Total tax:", np.sum(tax))

years_unique = np.unique(year)
for y in years_unique:
    print(int(y), np.sum(year == y))

for f in np.unique(fuel):
    print("Fuel", int(f), "avg mileage:", np.mean(mileage[fuel == f]))

print("Average MPG:", np.mean(mpg))

# second phase
def build_array(y):
    result = []
    for t in [1,2,3]:
        mask = (year == y) & (trans == t)
        avg_price = np.mean(price[mask])
        sum_mileage = np.sum(mileage[mask])
        max_tax = np.max(tax[mask])
        result.append([avg_price, sum_mileage, max_tax])
    return np.array(result)

arr2016 = build_array(2016)
arr2018 = build_array(2018)

difference = arr2018 - arr2016

def show_results(arr, year_value):
    print(f"\n{year_value} results")
    transmissions = ["Automatic", "Manual", "Semi-automatic"]

    for i in range(3):
        print(f"\nTransmission:", transmissions[i])
        print(f"Average Price : ${arr[i,0]:,.2f}")
        print(f"Sum Mileage   : {int(arr[i,1]):,}")
        print(f"Maximum Tax   : ${int(arr[i,2])}")

def show_difference(diff):
    print("\n(2018 - 2016) differemce")
    transmissions = ["Automatic", "Manual", "Semi-automatic"]

    for i in range(3):
        print(f"\nTransmission:", transmissions[i])
        print(f"Price Change   : ${diff[i,0]:,.2f}")
        print(f"Mileage Change : {int(diff[i,1]):,}")
        print(f"Tax Change     : ${int(diff[i,2])}")

show_results(arr2016, 2016)
show_results(arr2018, 2018)
show_difference(difference)
```
